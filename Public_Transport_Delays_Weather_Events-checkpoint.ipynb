```python
#Import libraries

# Step 1: Import Required Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
print("Project: Public Transport Delays with Weather and Events")
```

```python
#: Load dataset

# Step 2: Load Dataset

file_path = r"C:\Users\nishika tendane\Downloads\archive (12)\public_transport_delays.csv"

df = pd.read_csv(file_path)

print("Dataset Loaded Successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

display(df.head())
```

```python
#Check column names

# Step 3: Column Names

print("All Columns:")
print(df.columns.tolist())
```

```python
#Dataset information

# Step 4: Dataset Information

df.info()
```

```python
#Missing values

# Step 5: Missing Values

missing = df.isnull().sum()

print(missing[missing > 0])
print("Total Missing Values:", df.isnull().sum().sum())
```

```python
#Duplicate records

# Step 6: Duplicate Records

print("Duplicate Rows:", df.duplicated().sum())

df = df.drop_duplicates().copy()

print("Shape After Removing Duplicates:", df.shape)
```

```python
#Clean column names

# Step 7: Standardize Column Names

df.columns = (
    df.columns.str.strip()
    .str.lower()
    .str.replace(r"[^a-z0-9]+", "_", regex=True)
    .str.strip("_")
)

print(df.columns.tolist())
```

```python
#Understand numerical data

# Step 8: Numerical Summary

display(df.describe().T)
```

```python
#Understand categorical data

# Step 9: Categorical Summary

categorical_columns = df.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

print("Categorical Columns:", categorical_columns)

for col in categorical_columns:
    print(f"\n{col}")
    display(df[col].value_counts(dropna=False).head(10))
```

```python
#andle missing values

# Step 10: Handle Missing Values

for col in df.columns:
    if df[col].isnull().any():
        if pd.api.types.is_numeric_dtype(df[col]):
            df[col] = df[col].fillna(df[col].median())
        else:
            mode = df[col].mode(dropna=True)
            if not mode.empty:
                df[col] = df[col].fillna(mode.iloc[0])
            else:
                df[col] = df[col].fillna("Unknown")

print("Remaining Missing Values:", df.isnull().sum().sum())
```

```python
#Find possible delay column

# Step 11: Identify Delay-Related Columns

delay_columns = [
    col for col in df.columns
    if any(word in col for word in ["delay", "late", "minutes"])
]

print("Possible Delay Columns:", delay_columns)
```

```python
#Analyze delay values

# Step 12: Delay Summary

for col in delay_columns:
    print("\nAnalyzing:", col)

    if pd.api.types.is_numeric_dtype(df[col]):
        print(df[col].describe())
    else:
        print(df[col].value_counts().head(10))
```

```python
#Analyze weather-related columns

# Step 13: Weather Analysis

weather_columns = [
    col for col in df.columns
    if any(word in col for word in ["weather", "rain", "temperature", "wind", "snow"])
]

print("Weather Columns:", weather_columns)

for col in weather_columns:
    print("\n", col)
    display(df[col].value_counts().head(10))
```

```python
#Analyze event-related columns

# Step 14: Event Analysis

event_columns = [
    col for col in df.columns
    if any(word in col for word in ["event", "holiday", "festival", "crowd"])
]

print("Event Columns:", event_columns)

for col in event_columns:
    print("\n", col)
    display(df[col].value_counts().head(10))
```

```python
#Visualize delay distribution

# Step 15: Delay Distribution

numeric_delay_columns = [
    col for col in delay_columns
    if pd.api.types.is_numeric_dtype(df[col])
]

if numeric_delay_columns:
    col = numeric_delay_columns[0]

    df[col].plot(kind="hist", bins=20)
    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.show()
else:
    print("No numeric delay column found. Check Step 3 output.")
```

```python
#Compare delay with weather

# Step 16: Weather vs Delay

if weather_columns and numeric_delay_columns:
    weather_col = weather_columns[0]
    delay_col = numeric_delay_columns[0]

    display(
        df.groupby(weather_col, dropna=False)[delay_col]
        .agg(["count", "mean", "median"])
        .sort_values("mean", ascending=False)
    )
else:
    print("Suitable weather or numeric delay columns not found.")
```

```python
#Compare delay with events

# Step 17: Events vs Delay

if event_columns and numeric_delay_columns:
    event_col = event_columns[0]
    delay_col = numeric_delay_columns[0]

    display(
        df.groupby(event_col, dropna=False)[delay_col]
        .agg(["count", "mean", "median"])
        .sort_values("mean", ascending=False)
    )
else:
    print("Suitable event or numeric delay columns not found.")
```

```python
#Correlation analysis

# Step 18: Numerical Correlation

numeric_df = df.select_dtypes(include="number")
correlation = numeric_df.corr()

display(correlation)

if not correlation.empty:
    plt.figure(figsize=(10, 6))
    plt.imshow(correlation, aspect="auto")
    plt.colorbar()
    plt.xticks(
        range(len(correlation.columns)),
        correlation.columns,
        rotation=90
    )
    plt.yticks(range(len(correlation.columns)), correlation.columns)
    plt.title("Numerical Feature Correlation")
    plt.tight_layout()
    plt.show()
```

```python
#Save cleaned dataset

# Step 19: Save Cleaned Dataset

output_path = r"C:\Users\nishika tendane\Downloads\public_transport_delays_cleaned.csv"

df.to_csv(output_path, index=False)

print("Cleaned dataset saved successfully!")
print(output_path)
```
